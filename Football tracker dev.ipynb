```python
import pandas as pd
import matplotlib.pyplot as plt
import unicodedata

from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier

# =========================
# LOAD DATASETS
# =========================

match_df = pd.read_csv("champions_league_matches.csv")
player_df = pd.read_csv("attacking.csv")

# =========================
# CLEAN DATA
# =========================

match_df = match_df.dropna()

# =========================
# TRAIN MATCH PREDICTOR
# =========================

encoder = LabelEncoder()

all_teams = pd.concat([
    match_df["home_team"],
    match_df["away_team"]
])

encoder.fit(all_teams)

match_df["home_team_encoded"] = encoder.transform(match_df["home_team"])
match_df["away_team_encoded"] = encoder.transform(match_df["away_team"])

X = match_df[[
    "home_team_encoded",
    "away_team_encoded"
]]

y = match_df["result"]

model = DecisionTreeClassifier()
model.fit(X, y)

print("⚽ Football AI Ready!")

# =========================
# SMART TEXT CLEANER
# =========================

def clean_text(text):
    return ''.join(
        c for c in unicodedata.normalize('NFKD', str(text))
        if not unicodedata.combining(c)
    ).lower()

# =========================
# APP LOOP
# =========================

while True:

    query = input("\nType command (predict / player / assists / exit): ")

    # EXIT
    if query.lower() == "exit":
        print("👋 Goodbye!")
        break

    # =========================
    # MATCH PREDICTION
    # =========================
    elif query.lower().startswith("predict"):

        try:
            text = query.replace("predict", "").strip()
            home_name, away_name = text.split(" vs ")

            home_name = home_name.strip()
            away_name = away_name.strip()

            home_team = encoder.transform([home_name])[0]
            away_team = encoder.transform([away_name])[0]

            new_match = pd.DataFrame({
                "home_team_encoded": [home_team],
                "away_team_encoded": [away_team]
            })

            prediction = model.predict(new_match)

            print()

            if prediction[0] == 0:
                print(home_name, "Wins ⚽")
            elif prediction[0] == 1:
                print("Draw 🤝")
            else:
                print(away_name, "Wins ⚽")

        except:
            print("❌ Team not found in dataset")

    # =========================
    # PLAYER SEARCH (FIXED SMART VERSION)
    # =========================
    elif query.lower().startswith("player"):

        player_name = query.replace("player", "").strip()

        player = player_df[
            player_df["player_name"].apply(clean_text)
            .str.contains(clean_text(player_name), na=False)
        ]

        if player.empty:
            print("❌ Player not found")
        else:
            print(player.head())

    # =========================
    # TOP ASSISTS CHART
    # =========================
    elif query.lower() == "assists":

        top_players = player_df.sort_values(
            by="assists",
            ascending=False
        ).head(10)

        print(top_players[[
            "player_name",
            "assists"
        ]])

        top_players.plot(
            x="player_name",
            y="assists",
            kind="bar",
            figsize=(10,5)
        )

        plt.title("Top Assist Players")
        plt.show()

    else:
        print("Unknown command")
```
